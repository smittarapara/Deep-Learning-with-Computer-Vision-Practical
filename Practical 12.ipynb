### Aim

To perform data augmentation on the CIFAR-10 image dataset using different image transformation techniques and visualize the augmented images.

## Introduction

Data augmentation is a technique used in Deep Learning to artificially increase the size and diversity of a training dataset by applying different transformations to existing images.

It helps a CNN model learn variations in images and improves the model's ability to generalize to new and unseen data.

Common data augmentation techniques include:

- Rotation
- Horizontal Flipping
- Width Shifting
- Height Shifting
- Zooming
- Shearing

## CIFAR-10 Dataset

CIFAR-10 is a popular image classification dataset containing 60,000 color images belonging to 10 different classes.

### Dataset Information

- Total Images: 60,000
- Training Images: 50,000
- Testing Images: 10,000
- Image Size: 32 × 32 pixels
- Image Channels: 3 (RGB)
- Number of Classes: 10

### Classes

1. Airplane
2. Automobile
3. Bird
4. Cat
5. Deer
6. Dog
7. Frog
8. Horse
9. Ship
10. Truck

## Need for Data Augmentation

A deep learning model may perform poorly when the training dataset is small or when the training images have limited variations.

Data augmentation helps by:

- Increasing the diversity of training images.
- Reducing overfitting.
- Improving model generalization.
- Making the model more robust to image variations.
- Improving performance on unseen images.

## Data Augmentation Techniques

### 1. Rotation
Rotates an image by a specified angle.

### 2. Horizontal Flip
Flips the image from left to right.

### 3. Width Shift
Moves the image horizontally.

### 4. Height Shift
Moves the image vertically.

### 5. Zoom
Zooms into or out of the image.

These transformations generate new variations of existing training images without changing their class labels.

## Algorithm

1. Import the required libraries.
2. Load the CIFAR-10 dataset.
3. Select sample images from the training dataset.
4. Create an image data augmentation pipeline.
5. Apply different transformations to the images.
6. Generate augmented images.
7. Display the original and augmented images.
8. Compare the generated images with the original images.

```python
#Import Libraries
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras import datasets
from tensorflow.keras import layers
```

```python
# Load CIFAR-10 Dataset

(X_train, y_train), (X_test, y_test) = datasets.cifar10.load_data()

print("Training Images :", X_train.shape)
print("Testing Images  :", X_test.shape)
```

```python
#Define Class Names
class_names = [
    'airplane',
    'automobile',
    'bird',
    'cat',
    'deer',
    'dog',
    'frog',
    'horse',
    'ship',
    'truck'
]
```

```python
#Display Original Images
plt.figure(figsize=(10, 5))

for i in range(5):
    plt.subplot(1, 5, i + 1)
    plt.imshow(X_train[i])
    plt.title(class_names[y_train[i][0]])
    plt.axis('off')

plt.tight_layout()
plt.show()
```

## Creating Data Augmentation Pipeline

The Keras `Sequential` augmentation pipeline is used to apply different transformations to the training images.

In this practical, we use:

- Random Flip
- Random Rotation
- Random Zoom

```python
#Create Data Augmentation Pipeline
data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1)
])
```

```python
#Apply Augmentation
# Select one image

image = X_train[0]

# Add batch dimension

image = tf.expand_dims(image, axis=0)

# Generate augmented images

augmented_images = data_augmentation(image, training=True)

print("Original Image Shape :", image.shape)
print("Augmented Image Shape :", augmented_images.shape)
```

```python
#Display Augmented Images
plt.figure(figsize=(10, 5))

for i in range(5):
    
    augmented_image = data_augmentation(image, training=True)
    
    plt.subplot(1, 5, i + 1)
    plt.imshow(augmented_image[0].numpy().astype("uint8"))
    plt.title("Augmented")
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
#Compare Original and Augmented Images
plt.figure(figsize=(10, 4))

# Original Image
plt.subplot(1, 2, 1)
plt.imshow(X_train[0])
plt.title("Original Image")
plt.axis('off')

# Augmented Image
augmented_image = data_augmentation(image, training=True)

plt.subplot(1, 2, 2)
plt.imshow(augmented_image[0].numpy().astype("uint8"))
plt.title("Augmented Image")
plt.axis('off')

plt.show()
```

## Observation

- The original CIFAR-10 image is transformed using different augmentation techniques.
- Horizontal flipping changes the orientation of the image.
- Rotation slightly rotates the image.
- Zoom changes the scale of the image.
- The augmented images retain the original class label.
- Data augmentation provides additional variations of training images and can help reduce overfitting.

## Result

Data augmentation was successfully performed on the CIFAR-10 dataset using horizontal flipping, rotation, and zooming techniques. The generated images demonstrated how augmentation can create different variations of training images and help improve the generalization capability of deep learning models.
