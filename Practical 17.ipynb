### Aim

To perform image resizing and pixel normalization using OpenCV and visualize the original and processed images.

## Introduction

Image resizing and normalization are important preprocessing techniques used in Computer Vision and Deep Learning.

Image resizing changes the dimensions of an image to a specified width and height. It is commonly used to make all images in a dataset have the same dimensions.

Image normalization changes pixel values from their original range, usually 0–255, to a smaller range such as 0–1. This helps improve the stability and efficiency of machine learning and deep learning models.

## Theory

### 1. Image Resizing

Image resizing changes the spatial dimensions of an image.

OpenCV provides the `cv2.resize()` function for resizing images.

Syntax:

`cv2.resize(image, (width, height))`

For example, an image of size 640 × 480 pixels can be resized to 224 × 224 pixels.

### 2. Image Normalization

A digital image generally contains pixel values between 0 and 255.

Normalization converts these values into the range 0 to 1.

The formula is:

Normalized Pixel = Pixel Value / 255

For example:

- Pixel value 0 → 0.0
- Pixel value 128 → approximately 0.502
- Pixel value 255 → 1.0

## Importance of Resizing and Normalization

### Image Resizing

- Provides a fixed image size.
- Makes images suitable for CNN input.
- Reduces computational requirements.
- Ensures consistency across a dataset.

### Image Normalization

- Converts pixel values into a smaller range.
- Improves numerical stability.
- Helps neural networks train more efficiently.
- Makes image preprocessing consistent.

## Algorithm

1. Import OpenCV, NumPy, and Matplotlib.
2. Specify the path of the input image.
3. Read the image using `cv2.imread()`.
4. Display the original image and its dimensions.
5. Resize the image using `cv2.resize()`.
6. Normalize the resized image by dividing pixel values by 255.
7. Display the resized image.
8. Display the normalized image.
9. Save the resized image.
10. Display the minimum and maximum normalized pixel values.

```python
#Import Libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Read Image
# Specify image path

image = cv2.imread(r"C:\Users\Smit\Downloads\SOU LOGO.png")

# Read image

image = cv2.imread(r"C:\Users\Smit\Downloads\SOU LOGO.png")

if image is None:
    print("Image could not be loaded. Check the file path.")
else:
    print("Image loaded successfully.")
    print("Original Image Shape:", image.shape)
```

```python
#Display Original Image
# Convert BGR to RGB for Matplotlib

image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 4))

plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")

plt.show()
```

```python
#Resize Image
# Define new image size

new_width = 224
new_height = 224

# Resize image

resized_image = cv2.resize(
    image,
    (new_width, new_height)
)

print("Original Image Shape :", image.shape)
print("Resized Image Shape  :", resized_image.shape)
```

```python
#Display Resized Image
# Convert resized image from BGR to RGB

resized_rgb = cv2.cvtColor(
    resized_image,
    cv2.COLOR_BGR2RGB
)

plt.figure(figsize=(6, 4))

plt.imshow(resized_rgb)
plt.title("Resized Image (224 × 224)")
plt.axis("off")

plt.show()
```

## Image Normalization

The resized image contains pixel values ranging from 0 to 255.

To normalize the image, the pixel values are converted to floating-point values and divided by 255.

After normalization, the pixel values are in the range 0 to 1.

```python
#Normalize Image
# Convert image to float32

normalized_image = resized_image.astype(np.float32) / 255.0

print("Minimum Pixel Value:", normalized_image.min())
print("Maximum Pixel Value:", normalized_image.max())
```

```python
# Display normalized image

normalized_rgb = cv2.cvtColor(
    normalized_image,
    cv2.COLOR_BGR2RGB
)

plt.figure(figsize=(6, 4))

plt.imshow(normalized_rgb)
plt.title("Normalized Image")
plt.axis("off")

plt.show()
```

```python
#Compare Original and Resized Images
plt.figure(figsize=(10, 4))

# Original Image
plt.subplot(1, 2, 1)

plt.imshow(image_rgb)
plt.title(f"Original: {image.shape[1]} × {image.shape[0]}")
plt.axis("off")

# Resized Image
plt.subplot(1, 2, 2)

plt.imshow(resized_rgb)
plt.title("Resized: 224 × 224")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
#Save Resized Image
# Save resized image

output_path = "resized_image.jpg"

success = cv2.imwrite(
    output_path,
    resized_image
)

if success:
    print("Resized image saved successfully as:", output_path)
else:
    print("Image could not be saved.")
```
