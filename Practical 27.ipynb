# 27. To save and load trained deep learning models for reuse.

## Aim

To save and load trained deep learning models for reuse.

## Theory

A trained deep learning model can be saved to a file and loaded later for prediction or further training.

In this practical, a simple neural network is trained on the MNIST dataset, saved, and then loaded for reuse.

```python
#Import Libraries
import tensorflow as tf
from tensorflow.keras import layers, models

print("TensorFlow Version:", tf.__version__)
```

```python
#Load MNIST Dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

print("Dataset loaded successfully.")
```

```python
#Build Model
model = models.Sequential([
    layers.Flatten(input_shape=(28, 28)),
    layers.Dense(128, activation="relu"),
    layers.Dense(10, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
#Train Model
model.fit(
    x_train,
    y_train,
    epochs=2,
    batch_size=128,
    validation_split=0.1
)
```

```python
#Save Model
model.save("mnist_model.keras")

print("Model saved successfully.")
```

```python
#Load Model
loaded_model = tf.keras.models.load_model("mnist_model.keras")

print("Model loaded successfully.")
```

```python
#Evaluate Loaded Model
loss, accuracy = loaded_model.evaluate(x_test, y_test)

print("Test Accuracy:", accuracy)
```

```python
#Make Prediction
prediction = loaded_model.predict(x_test[:1], verbose=0)

predicted_class = tf.argmax(prediction[0]).numpy()

print("Predicted Digit:", predicted_class)
print("Actual Digit:", y_test[0])
```

## Result

The trained deep learning model was successfully saved, loaded, and reused for prediction on the MNIST dataset.
