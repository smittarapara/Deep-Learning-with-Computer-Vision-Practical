## Aim

To apply different image filtering and smoothing operations using OpenCV and observe their effects on an image.

## Introduction

Image filtering is an important image processing technique used to remove noise, smooth images, and improve image quality.

OpenCV provides several filtering and smoothing functions such as:

- Averaging Filter
- Gaussian Filter
- Median Filter
- Bilateral Filter

These filters modify the pixel values of an image based on the surrounding pixels.

## Theory

### 1. Averaging Filter

The averaging filter replaces each pixel with the average value of its neighboring pixels.

It is useful for reducing random noise and producing a smooth image.

OpenCV Function:

`cv2.blur()`

---

### 2. Gaussian Filter

Gaussian filtering uses a Gaussian kernel to smooth the image.

It provides better smoothing than a simple averaging filter and is commonly used before edge detection.

OpenCV Function:

`cv2.GaussianBlur()`

---

### 3. Median Filter

The median filter replaces each pixel with the median value of its neighboring pixels.

It is particularly effective for removing salt-and-pepper noise while preserving edges.

OpenCV Function:

`cv2.medianBlur()`

---

### 4. Bilateral Filter

The bilateral filter smooths an image while preserving important edges.

It considers both the spatial distance and intensity difference between pixels.

OpenCV Function:

`cv2.bilateralFilter()`

## Applications of Image Filtering

Image filtering and smoothing are used in:

- Noise removal
- Image enhancement
- Edge-preserving smoothing
- Preprocessing for computer vision
- Object detection
- Image segmentation
- Medical image processing
- Face and object recognition

## Algorithm

1. Import the required OpenCV and Matplotlib libraries.
2. Read the input image.
3. Convert the image from BGR to RGB for displaying with Matplotlib.
4. Apply the averaging filter.
5. Apply the Gaussian filter.
6. Apply the median filter.
7. Apply the bilateral filter.
8. Display the original and filtered images.
9. Save the filtered images.

```python
#Import Libraries
import cv2
import matplotlib.pyplot as plt
```

```python
# Read the image
image = cv2.imread(r"C:\Users\Smit\Downloads\SOU LOGO.png")

# Check whether image is loaded successfully
if image is None:
    print("Error: Image not found. Make sure 'sample.jpg' is in the same folder.")
else:
    print("Image loaded successfully.")
```

```python
#Convert Image for Display
# Convert BGR image to RGB
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Display original image
plt.figure(figsize=(6, 4))
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
# Apply averaging filter
average_filter = cv2.blur(image, (5, 5))

# Convert to RGB for display
average_rgb = cv2.cvtColor(average_filter, cv2.COLOR_BGR2RGB)

# Display
plt.figure(figsize=(6, 4))
plt.imshow(average_rgb)
plt.title("Averaging Filter")
plt.axis("off")
plt.show()
```

```python
#Apply Gaussian Filter
# Apply Gaussian Blur
gaussian_filter = cv2.GaussianBlur(image, (5, 5), 0)

# Convert to RGB
gaussian_rgb = cv2.cvtColor(gaussian_filter, cv2.COLOR_BGR2RGB)

# Display
plt.figure(figsize=(6, 4))
plt.imshow(gaussian_rgb)
plt.title("Gaussian Filter")
plt.axis("off")
plt.show()
```

```python
# Apply Median Filter
median_filter = cv2.medianBlur(image, 5)

# Convert to RGB
median_rgb = cv2.cvtColor(median_filter, cv2.COLOR_BGR2RGB)

# Display
plt.figure(figsize=(6, 4))
plt.imshow(median_rgb)
plt.title("Median Filter")
plt.axis("off")
plt.show()
```

```python
# Apply Bilateral Filter
bilateral_filter = cv2.bilateralFilter(image, 9, 75, 75)

# Convert to RGB
bilateral_rgb = cv2.cvtColor(bilateral_filter, cv2.COLOR_BGR2RGB)

# Display
plt.figure(figsize=(6, 4))
plt.imshow(bilateral_rgb)
plt.title("Bilateral Filter")
plt.axis("off")
plt.show()
```

```python
#Compare All Filters
plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(image_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(average_rgb)
plt.title("Averaging")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(gaussian_rgb)
plt.title("Gaussian")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(median_rgb)
plt.title("Median")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(bilateral_rgb)
plt.title("Bilateral")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
#Save Filtered Images
cv2.imwrite("average_filtered.jpg", average_filter)
cv2.imwrite("gaussian_filtered.jpg", gaussian_filter)
cv2.imwrite("median_filtered.jpg", median_filter)
cv2.imwrite("bilateral_filtered.jpg", bilateral_filter)

print("Filtered images saved successfully.")
```

| Gaussian Filter | Smooths the image effectively and reduces noise |
