### Aim

To read, display, and save images using OpenCV (Open Source Computer Vision Library) in Python.

## Introduction

OpenCV is an open-source computer vision library used for image and video processing. It provides various functions for reading, displaying, modifying, and saving images.

In this practical, OpenCV is used to perform three basic image operations:

- Read an image from a file.
- Display the image.
- Save the image to a specified location.

## Theory

An image can be represented as a matrix of pixel values. Each pixel contains information about the intensity or color of the image.

For a color image, OpenCV generally reads the image in **BGR (Blue, Green, Red)** format.

The three basic OpenCV functions used in this practical are:

### 1. `cv2.imread()`

Used to read an image from a specified file path.

### 2. `cv2.imshow()`

Used to display an image in a separate window.

### 3. `cv2.imwrite()`

Used to save an image to a specified file path.

## Common Image Reading Modes

OpenCV provides different modes for reading images:

- `cv2.IMREAD_COLOR` – Reads the image as a color image.
- `cv2.IMREAD_GRAYSCALE` – Reads the image as a grayscale image.
- `cv2.IMREAD_UNCHANGED` – Reads the image including the alpha channel, if available.

## Algorithm

1. Import the OpenCV library.
2. Specify the path of the input image.
3. Read the image using `cv2.imread()`.
4. Check whether the image has been loaded successfully.
5. Display the image.
6. Save the image using `cv2.imwrite()`.
7. Verify that the image has been saved successfully.

```python
#Import OpenCV
import cv2
import matplotlib.pyplot as plt
```

```python
#Read Image
# Specify image path

image = cv2.imread(r"C:\Users\Smit\Downloads\SOU LOGO.png")

if image is None:
    print("Image could not be loaded. Check the file path.")
else:
    print("Image loaded successfully.")
    print("Image Shape:", image.shape)
```

```python
#Display Image Using OpenCV
if image is not None:

    cv2.imshow("Original Image", image)

    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
#Display Image in Jupyter Notebook
if image is not None:

    # Convert BGR to RGB

    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    plt.imshow(image_rgb)
    plt.title("Original Image")
    plt.axis("off")
    plt.show()
```

## Why BGR to RGB Conversion is Required?

OpenCV reads color images in **BGR** format, while Matplotlib displays images using **RGB** format.

Therefore, `cv2.cvtColor()` is used to convert the image from BGR to RGB before displaying it using Matplotlib.

```python
#Save Image
# Specify output file path

output_path = "saved_image.jpg"

# Save image

success = cv2.imwrite(output_path, image)

if success:
    print("Image saved successfully as:", output_path)
else:
    print("Image could not be saved.")
```

```python
#Read Saved Image
saved_image = cv2.imread("saved_image.jpg")

if saved_image is not None:
    print("Saved image loaded successfully.")
    print("Saved Image Shape:", saved_image.shape)
```

```python
#Display Saved Image
saved_image_rgb = cv2.cvtColor(
    saved_image,
    cv2.COLOR_BGR2RGB
)

plt.imshow(saved_image_rgb)
plt.title("Saved Image")
plt.axis("off")
plt.show()
```

## Observation

- `cv2.imread()` successfully reads an image from the specified path.
- OpenCV reads color images in BGR format.
- `cv2.cvtColor()` converts BGR images to RGB for proper display using Matplotlib.
- `cv2.imwrite()` saves the image to the specified location.
- The saved image can be read again using `cv2.imread()`.
