# 19. To implement edge detection using Canny method in OpenCV.

### Aim

To implement edge detection using the Canny edge detection method in OpenCV and visualize the detected edges of a sample image.

## Introduction

Edge detection is an important technique in image processing and computer vision. It is used to identify boundaries and significant changes in intensity within an image.

Edges usually represent important features such as object boundaries, shapes, and contours.

The Canny Edge Detection method is one of the commonly used edge detection techniques because it can detect edges effectively while reducing the effect of noise.

## Theory

The Canny edge detection algorithm was developed to detect edges accurately in an image.

It generally involves the following steps:

1. Noise Reduction
2. Gradient Calculation
3. Non-Maximum Suppression
4. Double Thresholding
5. Edge Tracking by Hysteresis

OpenCV provides the `cv2.Canny()` function to perform Canny edge detection.

### Syntax

`cv2.Canny(image, threshold1, threshold2)`

Where:

- `image` is the input grayscale image.
- `threshold1` is the lower threshold.
- `threshold2` is the upper threshold.

Pixels with strong gradient values are identified as edges, while weak edges are retained or removed based on their connection to strong edges.

## Importance of Edge Detection

Edge detection is useful for:

- Object detection
- Object recognition
- Shape detection
- Image segmentation
- Contour detection
- Feature extraction
- Computer vision applications

## Algorithm

1. Import the required libraries.
2. Read the sample image using OpenCV.
3. Convert the color image into grayscale.
4. Apply Canny edge detection.
5. Specify lower and upper threshold values.
6. Display the original image.
7. Display the grayscale image.
8. Display the detected edges.
9. Save the edge-detected image.

```python
#Import Libraries
import cv2
import matplotlib.pyplot as plt
```

```python
#Read Sample Image
# Specify image path

image = cv2.imread(r"C:\Users\Smit\Downloads\SOU LOGO.png")

# Read image

image = cv2.imread(r"C:\Users\Smit\Downloads\SOU LOGO.png")

if image is None:
    print("Image could not be loaded. Check the file path.")
else:
    print("Image loaded successfully.")
    print("Image Shape:", image.shape)
```

```python
# Convert BGR image to grayscale

gray_image = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2GRAY
)

print("Image converted to grayscale successfully.")
```

```python
#Display Grayscale Image
plt.figure(figsize=(6, 4))

plt.imshow(gray_image, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.show()
```

## Applying Canny Edge Detection

The Canny method requires two threshold values:

- Lower Threshold = 100
- Upper Threshold = 200

These values are used to determine strong and weak edges in the image.

```python
# Apply Canny Edge Detection

lower_threshold = 100
upper_threshold = 200

edges = cv2.Canny(
    gray_image,
    lower_threshold,
    upper_threshold
)

print("Canny edge detection applied successfully.")
```

```python
#Display Detected Edges
plt.figure(figsize=(6, 4))

plt.imshow(edges, cmap="gray")
plt.title("Canny Edge Detection")
plt.axis("off")

plt.show()
```

```python
#Compare Original and Edge Image
# Convert original image from BGR to RGB

image_rgb = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2RGB
)

plt.figure(figsize=(10, 4))

# Original Image
plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")

# Canny Edge Image
plt.subplot(1, 2, 2)
plt.imshow(edges, cmap="gray")
plt.title("Canny Edges")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
#Save Edge-Detected Image
# Save Canny edge image

output_path = "canny_edges.jpg"

success = cv2.imwrite(
    output_path,
    edges
)

if success:
    print("Edge-detected image saved successfully as:", output_path)
else:
    print("Image could not be saved.")
```

## Effect of Threshold Values

The lower and upper threshold values affect the number of edges detected.

- Lower threshold detects more possible edges.
- Higher threshold detects stronger edges.
- Very low thresholds may detect unwanted noise.
- Very high thresholds may miss some important edges.

Therefore, suitable threshold values should be selected according to the image.

```python
#Experiment with Different Thresholds
# Apply Canny with different threshold values

edges_1 = cv2.Canny(gray_image, 50, 150)
edges_2 = cv2.Canny(gray_image, 100, 200)
edges_3 = cv2.Canny(gray_image, 150, 250)

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(edges_1, cmap="gray")
plt.title("Threshold: 50, 150")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(edges_2, cmap="gray")
plt.title("Threshold: 100, 200")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(edges_3, cmap="gray")
plt.title("Threshold: 150, 250")
plt.axis("off")

plt.tight_layout()
plt.show()
```

## Observation

- The Canny method successfully detects important edges in the image.
- Converting the image to grayscale is used before applying Canny edge detection.
- The detected edges represent boundaries and significant intensity changes.
- Changing the threshold values changes the number of detected edges.
- Appropriate threshold values help reduce unwanted edges and preserve important features.
