# 24. To develop an object detection system using a pre-trained YOLO model.

## Aim

To develop an object detection system using a pre-trained YOLO model.

## Theory

YOLO (You Only Look Once) is a deep learning-based object detection algorithm.

It can detect multiple objects in an image and identify their class along with their bounding boxes.

In this practical, a pre-trained YOLO model is used for object detection.

```python
#Import Library
from ultralytics import YOLO
from PIL import Image
import matplotlib.pyplot as plt
```

```python
#Load Pre-trained YOLO Model
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

print("YOLO model loaded successfully.")
```

```python
#Load Image
import urllib.request
from PIL import Image
import matplotlib.pyplot as plt

image_url = "https://ultralytics.com/images/bus.jpg"
image_path = "bus.jpg"

urllib.request.urlretrieve(image_url, image_path)

image = Image.open(image_path)

plt.imshow(image)
plt.axis("off")
plt.show()
```

```python
#Object Detection
results = model(image_path)

print("Object detection completed.")
```

```python
#Display Result
result_image = results[0].plot()

plt.figure(figsize=(10, 6))
plt.imshow(result_image)
plt.axis("off")
plt.show()
```

```python
#Display Detected Objects
for result in results:
    for box in result.boxes:
        class_id = int(box.cls[0])
        confidence = float(box.conf[0])

        print(
            "Object:", model.names[class_id],
            "| Confidence:", round(confidence, 2)
        )
```

## Result

An object detection system was successfully developed using a pre-trained YOLO model to detect and classify objects in an image.
