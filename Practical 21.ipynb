## Introduction

Face detection is a computer vision technique used to identify human faces in an image or video.

OpenCV provides a pre-trained Haar Cascade classifier that can detect faces efficiently.

The Haar Cascade classifier uses a set of simple features and a cascade of classifiers to identify objects such as faces.

In this practical, we will:
- Load an input image.
- Convert the image into grayscale.
- Load the pre-trained Haar Cascade face classifier.
- Detect faces in the image.
- Draw rectangles around detected faces.
- Display and save the result.

## Theory

### Haar Cascade Classifier

Haar Cascade is a machine-learning-based object detection method.

It was introduced for efficient object detection and is commonly used for detecting faces.

OpenCV provides pre-trained Haar Cascade XML files for different detection tasks.

For face detection, we use:

`haarcascade_frontalface_default.xml`

The classifier works by examining different regions of an image and identifying patterns that are likely to represent a human face.

### Main Steps

1. Load the Haar Cascade classifier.
2. Read the input image.
3. Convert the image into grayscale.
4. Apply the `detectMultiScale()` function.
5. Identify the coordinates of detected faces.
6. Draw rectangles around the detected faces.
7. Display and save the output image.

## Important OpenCV Function

### `detectMultiScale()`

The `detectMultiScale()` function is used to detect objects of different sizes in an image.

Syntax:

`detectMultiScale(image, scaleFactor, minNeighbors)`

### Parameters

- **image** – Grayscale input image.
- **scaleFactor** – Specifies how much the image size is reduced at each scale.
- **minNeighbors** – Specifies how many neighboring detections are required to retain a detected face.

Example:

`faces = face_cascade.detectMultiScale(gray, 1.1, 5)`

```python
#Import Libraries
import cv2
import matplotlib.pyplot as plt
```

```python
#Load Haar Cascade Classifier
# Load the pre-trained Haar Cascade classifier
face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)

print("Haar Cascade classifier loaded successfully.")
```

```python
# Read the input image
image = cv2.imread(r"C:\Users\Smit\Downloads\ICT-SCT\WhatsApp Image 2026-08-19 at 8.05.53 AM.jpeg")

if image is None:
    print("Error: Image not found. Make sure 'sample.jpg' is in the same folder.")
else:
    print("Image loaded successfully.")
```

```python
#Convert Image to Grayscale
# Convert BGR image to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Display grayscale image
plt.figure(figsize=(7, 5))
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
#Detect Faces
# Detect faces
faces = face_cascade.detectMultiScale(
    gray,
    scaleFactor=1.1,
    minNeighbors=5,
    minSize=(30, 30)
)

print("Number of faces detected:", len(faces))
```

```python
#Draw Rectangle Around Faces
# Draw rectangles around detected faces
result = image.copy()

for (x, y, w, h) in faces:
    cv2.rectangle(
        result,
        (x, y),
        (x + w, y + h),
        (0, 255, 0),
        2
    )

# Convert BGR to RGB for Matplotlib
result_rgb = cv2.cvtColor(result, cv2.COLOR_BGR2RGB)

# Display result
plt.figure(figsize=(8, 6))
plt.imshow(result_rgb)
plt.title("Face Detection Result")
plt.axis("off")
plt.show()
```

```python
#Display Original and Detected Image
# Convert original image to RGB
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(result_rgb)
plt.title("Detected Faces")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
#Save Output Image
# Save the face detection result
cv2.imwrite("face_detection_result.jpg", result)

print("Face detection result saved as 'face_detection_result.jpg'.")
```
