## Introduction

Image thresholding is a fundamental image processing technique used to separate objects or regions of an image from its background.

Thresholding converts a grayscale image into a binary image by comparing each pixel value with a predefined threshold value.

Thresholding is commonly used in:

- Object detection
- Image segmentation
- Character recognition
- Document processing
- Shape detection

## Theory

In basic thresholding, each pixel is compared with a threshold value.

If the pixel value is greater than the threshold, it is assigned one value, and otherwise it is assigned another value.

OpenCV provides the `cv2.threshold()` function for applying thresholding.

### Syntax

`cv2.threshold(image, threshold_value, maximum_value, threshold_type)`

### Common Thresholding Techniques

1. Binary Thresholding
2. Binary Inverse Thresholding
3. Truncated Thresholding
4. Threshold to Zero
5. Threshold to Zero Inverse

## Types of Thresholding

### 1. Binary Thresholding

Pixels greater than the threshold are assigned the maximum value, while other pixels become zero.

### 2. Binary Inverse

The result of binary thresholding is inverted.

### 3. Truncated Thresholding

Pixel values above the threshold are set to the threshold value.

### 4. Threshold to Zero

Pixel values below the threshold are set to zero, while values above the threshold remain unchanged.

### 5. Threshold to Zero Inverse

Pixel values above the threshold are set to zero, while values below the threshold remain unchanged.

```python
#Import Libraries
import cv2
import matplotlib.pyplot as plt
```

```python
#Read Sample Image
# Specify image path

image = cv2.imread(r"C:\Users\Smit\Downloads\SOU LOGO.png")

# Read image

image = cv2.imread(r"C:\Users\Smit\Downloads\SOU LOGO.png")

if image is None:
    print("Image could not be loaded. Check the file path.")
else:
    print("Image loaded successfully.")
    print("Image Shape:", image.shape)
```

```python
# Convert color image to grayscale

gray_image = cv2.cvtColor(
    image,
    cv2.COLOR_BGR2GRAY
)

print("Image converted to grayscale successfully.")
```

```python
#Display Grayscale Image
plt.figure(figsize=(6, 4))

plt.imshow(gray_image, cmap="gray")
plt.title("Original Grayscale Image")
plt.axis("off")

plt.show()
```

## Applying Thresholding

A threshold value of 127 is used for demonstrating the different thresholding techniques.

The maximum pixel value is set to 255.

```python
# Binary Thresholding

ret, binary = cv2.threshold(
    gray_image,
    127,
    255,
    cv2.THRESH_BINARY
)

print("Binary Thresholding Applied")
```

```python
# Binary Inverse Thresholding

ret, binary_inverse = cv2.threshold(
    gray_image,
    127,
    255,
    cv2.THRESH_BINARY_INV
)

print("Binary Inverse Thresholding Applied")
```

```python
# Truncated Thresholding

ret, truncated = cv2.threshold(
    gray_image,
    127,
    255,
    cv2.THRESH_TRUNC
)

print("Truncated Thresholding Applied")
```

```python
# Threshold to Zero

ret, threshold_zero = cv2.threshold(
    gray_image,
    127,
    255,
    cv2.THRESH_TOZERO
)

print("Threshold to Zero Applied")
```

```python
# Threshold to Zero Inverse

ret, threshold_zero_inverse = cv2.threshold(
    gray_image,
    127,
    255,
    cv2.THRESH_TOZERO_INV
)

print("Threshold to Zero Inverse Applied")
```

```python
#Display All Results
plt.figure(figsize=(12, 8))

# Original
plt.subplot(2, 3, 1)
plt.imshow(gray_image, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

# Binary
plt.subplot(2, 3, 2)
plt.imshow(binary, cmap="gray")
plt.title("Binary")
plt.axis("off")

# Binary Inverse
plt.subplot(2, 3, 3)
plt.imshow(binary_inverse, cmap="gray")
plt.title("Binary Inverse")
plt.axis("off")

# Truncated
plt.subplot(2, 3, 4)
plt.imshow(truncated, cmap="gray")
plt.title("Truncated")
plt.axis("off")

# Threshold to Zero
plt.subplot(2, 3, 5)
plt.imshow(threshold_zero, cmap="gray")
plt.title("Threshold to Zero")
plt.axis("off")

# Threshold to Zero Inverse
plt.subplot(2, 3, 6)
plt.imshow(threshold_zero_inverse, cmap="gray")
plt.title("Threshold to Zero Inverse")
plt.axis("off")

plt.tight_layout()
plt.show()
```

## Adaptive Thresholding

Adaptive thresholding calculates the threshold value for smaller regions of an image instead of using one fixed threshold value.

It is useful when an image has uneven lighting or different brightness levels.

OpenCV provides two common adaptive thresholding methods:

- Adaptive Mean Thresholding
- Adaptive Gaussian Thresholding

```python
# Adaptive Mean Thresholding

adaptive_mean = cv2.adaptiveThreshold(
    gray_image,
    255,
    cv2.ADAPTIVE_THRESH_MEAN_C,
    cv2.THRESH_BINARY,
    11,
    2
)

plt.figure(figsize=(6, 4))

plt.imshow(adaptive_mean, cmap="gray")
plt.title("Adaptive Mean Thresholding")
plt.axis("off")

plt.show()
```

```python
# Adaptive Gaussian Thresholding

adaptive_gaussian = cv2.adaptiveThreshold(
    gray_image,
    255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY,
    11,
    2
)

plt.figure(figsize=(6, 4))

plt.imshow(adaptive_gaussian, cmap="gray")
plt.title("Adaptive Gaussian Thresholding")
plt.axis("off")

plt.show()
```

```python
# Save Binary Thresholded Image

output_path = "thresholded_image.jpg"

success = cv2.imwrite(
    output_path,
    binary
)

if success:
    print("Thresholded image saved successfully as:", output_path)
else:
    print("Image could not be saved.")
```

- Truncated thresholding limits pixel values above the selected threshold.
- Threshold-to-zero preserves pixels above the threshold.
- Adaptive thresholding is useful for images with uneven lighting.
