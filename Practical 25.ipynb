# 25. To implement image segmentation using deep learning models.			

## Aim

To implement image segmentation using a deep learning model.

## Theory

Image segmentation divides an image into different regions or objects.

In this practical, a pre-trained YOLO segmentation model is used to identify and segment objects in an image.

```python
#Install YOLO
!pip install ultralytics
```

```python
#Import Libraries
from ultralytics import YOLO
import urllib.request
from PIL import Image
import matplotlib.pyplot as plt
```

```python
#Load Pre-trained Segmentation Model
model = YOLO("yolov8n-seg.pt")

print("YOLO segmentation model loaded successfully.")
```

```python
#Download and Load Image
image_url = "https://ultralytics.com/images/bus.jpg"
image_path = "bus.jpg"

urllib.request.urlretrieve(image_url, image_path)

image = Image.open(image_path)

plt.imshow(image)
plt.axis("off")
plt.show()
```

```python
#Perform Image Segmentation
results = model(image_path)

print("Image segmentation completed.")
```

```python
#Display Segmentation Result
result_image = results[0].plot()

plt.figure(figsize=(10, 6))
plt.imshow(result_image)
plt.axis("off")
plt.show()
```

```python
#Display Segmented Objects
if results[0].masks is not None:
    print("Number of segmented objects:", len(results[0].masks))
else:
    print("No objects were segmented.")
```

## Result

Image segmentation was successfully performed using a pre-trained YOLO deep learning model.
