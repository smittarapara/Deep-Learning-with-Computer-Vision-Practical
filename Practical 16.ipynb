# 16. To convert color images into grayscale images using OpenCV.

### Aim

To convert a color image into a grayscale image using OpenCV and display the original and grayscale images.

## Introduction

A color image contains three color channels: Red, Green, and Blue (RGB). Each pixel in a color image contains information about these three channels.

A grayscale image contains only intensity information. Instead of three color channels, it represents each pixel using a single intensity value ranging from black to white.

Grayscale conversion is commonly used in computer vision because it reduces the amount of image data and makes many image processing operations simpler.

## Theory

OpenCV provides the `cv2.cvtColor()` function to convert an image from one color space to another.

For converting a color image into grayscale, the following syntax is used:

`cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)`

OpenCV reads color images in **BGR (Blue, Green, Red)** format by default.

The grayscale image is generated using a weighted combination of the three color channels. Green contributes more to perceived brightness than red and blue.

## Advantages of Grayscale Images

- Reduces image complexity.
- Requires less storage than a color image.
- Reduces computational requirements.
- Useful for edge detection and image segmentation.
- Commonly used as a preprocessing step in computer vision and Deep Learning.

## Algorithm

1. Import the OpenCV and Matplotlib libraries.
2. Specify the path of the color image.
3. Read the color image using `cv2.imread()`.
4. Convert the BGR image into grayscale using `cv2.cvtColor()`.
5. Display the original color image.
6. Display the grayscale image.
7. Save the grayscale image using `cv2.imwrite()`.

```python
#Import Libraries
import cv2
import matplotlib.pyplot as plt
```

```python
# Read Color Image
# Specify image path

image = cv2.imread(r"C:\Users\Smit\Downloads\SOU LOGO.png")

# Read the color image

image = cv2.imread(r"C:\Users\Smit\Downloads\SOU LOGO.png")

if image is None:
    print("Image could not be loaded. Check the file path.")
else:
    print("Color image loaded successfully.")
    print("Image Shape:", image.shape)
```

```python
# Convert BGR image to grayscale

gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

print("Image successfully converted to grayscale.")
print("Grayscale Image Shape:", gray_image.shape)
```

```python
#Display Original Color Image
# Convert BGR to RGB for Matplotlib

image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 4))

plt.imshow(image_rgb)
plt.title("Original Color Image")
plt.axis("off")

plt.show()
```

```python
#Display Grayscale Image
plt.figure(figsize=(6, 4))

plt.imshow(gray_image, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.show()
```

```python
#Display Both Images
plt.figure(figsize=(10, 4))

# Original image
plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title("Original Color Image")
plt.axis("off")

# Grayscale image
plt.subplot(1, 2, 2)
plt.imshow(gray_image, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
#Save Grayscale Image
# Save grayscale image

output_path = "grayscale_image.jpg"

success = cv2.imwrite(output_path, gray_image)

if success:
    print("Grayscale image saved successfully as:", output_path)
else:
    print("Image could not be saved.")
```

```python
#Verify Saved Image
# Read the saved grayscale image

saved_image = cv2.imread(
    "grayscale_image.jpg",
    cv2.IMREAD_GRAYSCALE
)

if saved_image is not None:
    print("Saved grayscale image loaded successfully.")
    print("Saved Image Shape:", saved_image.shape)
```

## Observation

- The color image contains three channels.
- The grayscale image contains a single intensity channel.
- `cv2.cvtColor()` successfully converts the color image into grayscale.
- Grayscale conversion removes color information while preserving image intensity.
- The grayscale image requires less data than the original color image.
